```python
import numpy as np
import pandas as pd

df2 = pd.DataFrame([
    {"name": "alice", "age": 25, "score": 88.5},
    {"name": "bob", "age": 30, "score": 92.0},
    {"name": "charlie", "age": 35, "score": 75.0}
])

print(df2)
print(df2["age"].mean())
print(df2.columns)
print(df2.describe)
print(df2.info)
print(df2.index)
df2.loc[1,"age"]=27
print(df2)
df2["age_next_year"]=df2["age"]+1
df2["adult"]=df2["age"]>=18
df2["grade"]=df2["score"].apply(
    lambda  x: "A" if x>=90 else "B" if x>=80 else "C"
)
print(df2)
df2=df2.drop(columns=["age_next_year"])
print(df2)
```

```python
import numpy as np
import pandas as pd
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_excel(filename)
print(df)
df["adult"] = df["age"] >= 18
print("Mean score:", df["score"].mean())
print("Sum of ages:", df["age"].sum())
print("Standard deviation:", df["score"].std())
grouped = df.groupby("adult")["score"].mean()
print("\nAverage score by adult:")
print(grouped)
agg_df = df.groupby("adult").agg(
    avg_score=("score", "mean"),
    max_age=("age", "max"),
    count=("name", "size")
).reset_index()
print("\nAggregated DataFrame:")
print(agg_df)
```
